```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
documents = [
    "This is a list which contains sample documents",
    "Keywords are important for keyword-based search",
    "Document analysis involves extracting keywords",
    "Keyword-based search relies on sparse embeddings"
]
```

```python
query = "Keyword Based Search"
```

```python
import re

def preprocess_text(text):
    # Convert text to lowercase
    text = text.lower()
    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)
    return text 
```

```python
preprocess_documents = [preprocess_text(doc) for doc in documents]
print("Pre-processed Documents")
print(preprocess_documents)

preprocess_query = preprocess_text(query)
print("Pre-processed Query")
print(preprocess_query)
```

```python
vectorizer = TfidfVectorizer()
```

```python
doc_vectors = vectorizer.fit_transform(preprocess_documents)
```

```python
doc_vectors = doc_vectors.toarray()
doc_vectors[0]
```

```python
query_vector = vectorizer.transform([preprocess_query])
```

```python
query_vector = query_vector.toarray()
```

```python
query_vector
```

```python
similarities = cosine_similarity(doc_vectors, query_vector)
```

```python
similarities
```

```python
# Ranking
rank_indices = np.argsort(similarities, axis=0)[::-1].flatten()
```

```python
rank_indices
```

```python
ranked_documents = [documents[i] for i in rank_indices]
```

```python
for i, doc in enumerate(ranked_documents):
    print(f"Rank {i+1}: {doc}")
```
